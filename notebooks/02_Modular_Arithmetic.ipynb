# 02. Modular Arithmetic

This notebook covers congruences, modular arithmetic, and modular inverse.

$$a \equiv b \pmod{m}$$
means that $a-b$ is divisible by $m$.

```python
import sys
from pathlib import Path

root = Path.cwd().resolve()
if (root / 'src').exists():
    sys.path.insert(0, str(root))
else:
    sys.path.insert(0, str(root.parent))

from src.modular_arithmetic import mod_add, mod_sub, mod_mul, mod_pow, mod_inverse
```

```python
print('(7 + 5) mod 11 =', mod_add(7, 5, 11))
print('(7 - 5) mod 11 =', mod_sub(7, 5, 11))
print('(7 * 5) mod 11 =', mod_mul(7, 5, 11))
print('2^10 mod 13 =', mod_pow(2, 10, 13))
print('Inverse of 3 mod 11 =', mod_inverse(3, 11))
```

## Explanation

The modular inverse exists only when gcd(a, m) = 1. This is essential in RSA and many cryptographic methods.
