This notebook introduces Caesar and affine ciphers to show the relationship between number theory and cryptography.

These examples are educational only and are not secure production systems.

```python
import sys
from pathlib import Path

root = Path.cwd().resolve()
if (root / 'src').exists():
    sys.path.insert(0, str(root))
else:
    sys.path.insert(0, str(root.parent))

from src.cryptography import caesar_encrypt, caesar_decrypt, affine_encrypt, affine_decrypt
```

```python
plain = 'HELLO'
cipher = caesar_encrypt(plain, 3)
decoded = caesar_decrypt(cipher, 3)
print('Plain:', plain)
print('Caesar:', cipher)
print('Decoded:', decoded)

affine_text = affine_encrypt('NUMBER', 5, 8)
print('Affine cipher:', affine_text)
print('Affine decode:', affine_decrypt(affine_text, 5, 8))
```

## Explanation

Classical ciphers rely on simple arithmetic modulo 26. Even though they are easy to understand, they are not secure against modern attacks.
