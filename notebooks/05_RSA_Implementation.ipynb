# 05. RSA Implementation

This notebook demonstrates a small educational RSA example using integer arithmetic and modular exponentiation.

$$c \equiv m^e \pmod n$$

$$m \equiv c^d \pmod n$$

```python
import sys
from pathlib import Path

root = Path.cwd().resolve()
if (root / 'src').exists():
    sys.path.insert(0, str(root))
else:
    sys.path.insert(0, str(root.parent))

from src.cryptography import generate_rsa_keypair, rsa_encrypt, rsa_decrypt
```

```python
keys = generate_rsa_keypair(50, 200)
message = 42
cipher = rsa_encrypt(message, keys['e'], keys['n'])
recovered = rsa_decrypt(cipher, keys['d'], keys['n'])

print('p =', keys['p'])
print('q =', keys['q'])
print('n =', keys['n'])
print('phi(n) =', keys['phi_n'])
print('e =', keys['e'])
print('d =', keys['d'])
print('Message =', message)
print('Ciphertext =', cipher)
print('Recovered =', recovered)
```

## Explanation

The example confirms that RSA encryption and decryption are inverses modulo n when the keys are valid. This is a classroom illustration, not a production-ready security implementation.
